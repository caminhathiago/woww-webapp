```python
from woww.api.erddap import WOWWErdappData
from datetime import datetime, timedelta

%load_ext autoreload
%autoreload 2
```

```python
wd = WOWWErdappData(server="https://pae-paha.pacioos.hawaii.edu/erddap/",
                    protocol="griddap",
                    dataset_id="ww3_global",
                    initialize=True)
```

```python
wd.set_vars_constraints(variables=['Tdir', 'Tper', 'Thgt',
                                    'sdir','sper','shgt',
                                    'wdir','wper','whgt'],
                        longitude=115,#(305,333),
                        latitude=-31,#(-35,4),
                        start_date=(datetime.now()-timedelta(days=1)).strftime("%Y-%m-%dT%H:%M:%SZ"),
                        correct_pos=False)

ww3 = wd.grab_batch_data(response_type="pandas")
```

```python
#TRANSFORM
ww3 = wd.process_var_labels(ww3)
ww3 = wd.rename_var_labels(ww3)
```

```python
ww3['tdir_windrose'] = wd.get_direc_quadrant(data=ww3,direc_var='tdir')
ww3['sdir_windrose'] = wd.get_direc_quadrant(data=ww3,direc_var='sdir')
ww3['wdir_windrose'] = wd.get_direc_quadrant(data=ww3,direc_var='wdir')
```

```python
wd2 = WOWWErdappData(server="https://pae-paha.pacioos.hawaii.edu/erddap/",
            protocol="griddap",
            dataset_id="ncep_global")
```

```python
wd2.set_vars_constraints(variables=['ugrd10m', 'vgrd10m'],
                        longitude=115,
                        latitude=-31,
                start_date=(datetime.now()-timedelta(days=1)).strftime("%Y-%m-%dT%H:%M:%SZ"),
                correct_pos=False)

gfs = wd2.grab_batch_data(response_type="pandas")

gfs = wd2.process_var_labels(gfs)
gfs = wd2.rename_var_labels(gfs)
gfs = wd2.select_vars(gfs)
gfs['wspd'] = wd2.calc_wind_veloc(gfs)
gfs['wdir'] = wd2.calc_wind_direc(gfs)
```
